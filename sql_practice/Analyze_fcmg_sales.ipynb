```python
# Level 1 — Filtering & Selection

# . High-value sales

# Find all transactions where:

# Revenue is greater than 1,000.

query = '''
SELECT Invoice_ID, City, Category,Brand, Revenue
FROM sales
WHERE Revenue > 1000
'''
print(pd.read_sql_query(query, conn))
```

```python
# 2. Snacks in selected cities

# Find transactions where:

# Category is 'Snacks'
# AND the city is either Kolkata, Delhi, or Mumbai

query = '''
SELECT City, Category,Brand, Units, Revenue
FROM sales
WHERE Category = "Snacks"  AND City IN ('Kolkata', 'Delhi', 'Mumbai')
'''
print(pd.read_sql_query(query, conn))
```
