```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("Parvati_Secondary_School.db")
```

```python
cursor = conn.cursor()
```

```python
cursor.execute('''
CREATE TABLE courses (
    course_id INTEGER PRIMARY KEY,
    course_name TEXT NOT NULL,
    duration INTEGER,
    fee REAL
)
''')
```

```python
cursor.execute('''
CREATE TABLE students (
    student_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    gender TEXT,
    age INTEGER,
    city TEXT,
    course_id INTEGER,
    marks REAL,
    email TEXT,
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
) 
''')
```

```python
print("Table Created Successfully!")
```

```python
print("Students name, marks and updated marks after adding 10 to each marks ")
print(pd.read_sql_query('''SELECT name, marks, marks + 10 AS updated_marks
FROM students''',conn))
```

```python
print('''Displaying the name, city, marks, and age of students who
are from Kathmandu OR Lalitpur
AND have marks greater than or equal to 85
AND are not 21 years old''')
print(pd.read_sql_query('''SELECT name, city, marks, age 
FROM students
WHERE (city = "Kathmandu" OR city = "Lalitpur") AND (marks >= 85) AND (age != 21)''',conn))
```

```python

print('''Displaying name, gender, city, and marks of students who
are from Kathmandu, Pokhara, or Bhaktapur
AND their marks are between 80 and 95, inclusive
AND their gender is not Male''')
print(pd.read_sql_query('''SELECT name, gender, city, marks 
FROM students
WHERE city IN ("Kathmandu", "Pokhara", "Bhaktapur") AND marks BETWEEN 80 AND 95 AND gender != "Male"''',conn))
```

```python
print('''Displaying name, city, marks, and email of students who satisfy either: 
Their name starts with A
AND their marks are greater than 85
OR
Their name ends with a
AND they are from Kathmandu or Lalitpur
AND their marks are less than 80''')
print(pd.read_sql_query('''SELECT name, city, marks, email
FROM students
WHERE (name LIKE "A%" AND marks > 85) OR (name LIKE "%a" AND city IN ("Kathmandu", "Lalitpur") AND marks < 80)''',conn))
```

```python

print("From the students table, displaying all the different cities represented by the students.")
print(pd.read_sql_query('''SELECT DISTINCT city 
FROM students''',conn))
```

```python

print("Displaying name and marks where the students marks are sorted from highest marks to lowest marks")
print(pd.read_sql_query('''SELECT name, marks 
FROM students
ORDER BY marks DESC''',conn))
```

```python
print('''Displaying name, city, age, marks and sorting the result by
city alphabetically (A → Z)
For students in the same city, sorting by marks from highest to lowest''')


print(pd.read_sql_query('''SELECT name, age, city, marks 
FROM students
ORDER BY city, marks DESC''',conn))
```

```python
print("7 first Students:")
print(pd.read_sql_query('''SELECT *
FROM students 
LIMIT 7''',conn))
print("Top 5 students:")
print(pd.read_sql_query('''
SELECT name, marks 
FROM students
ORDER BY marks DESC
LIMIT 5''',conn))

print("Top 3 young students:")
print(pd.read_sql_query('''
SELECT name, age, city
FROM students
ORDER BY age
LIMIT 3''',conn))
```

```python

print('''Displaying 
total number of students
total marks
average marks
highest marks
lowest marks''')
print(pd.read_sql_query('''
SELECT COUNT(*) AS total_student_count, SUM(marks) AS total_marks, AVG(marks) AS average_marks, MAX(marks) AS max_marks, MIN(marks) AS min_marks
FROM students''',conn))
```

```python

print('''Display each city, number of students and average marks of each city: ''')
print(pd.read_sql_query('''
SELECT city, COUNT(*) AS no_of_students,AVG(marks) as average_marks
FROM students
GROUP BY city''',conn))
```

```python
print("Displaying the course_id and the average marks for each course, but showing only courses whose average marks are greater than 85")
print(pd.read_sql_query('''
SELECT course_id, AVG(marks) AS average_marks 
FROM students
GROUP BY course_id 
HAVING AVG(marks) > 85''',conn))
```
